```python
# Importing pandas library as pd for data manipulation
import pandas as pd
# Importing StandardScaler from sklearn.preprocessing for feature scaling
from sklearn.preprocessing import StandardScaler

# Set file path to the location of the CSV file
file_path = r'C:\Users\lenovo\Desktop\ML-Zeolite-Selective\data\2_unsupervised_learning\2_input\unsupervised_learning.csv'

# Reading the CSV file into a DataFrame using pandas
df = pd.read_csv(file_path, encoding='utf-8')
# Display the DataFrame to inspect its contents
df
```

```python
# Select the 'Framework' column from the DataFrame df and assign it to the variable framework_col.
framework_col = df['Framework']

# Create a new DataFrame df1 that consists of all columns from the third column onwards (excluding the first two columns) of the original DataFrame df.
df1 = df.iloc[:, 1:]

# Display the DataFrame df1, which now contains only the columns from the third onward.
df1
```

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Standardize the data in df1 (features will be centered around 0 with a standard deviation of 1)
df1_standardized = scaler.fit_transform(df1)

# Convert the standardized data back to a DataFrame, keeping the original column names from df1
df1_standardized = pd.DataFrame(df1_standardized, columns=df1.columns)

# Display the resulting DataFrame df1_standardized, which now contains standardized data
df1_standardized
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Initialize PCA without specifying the number of components; it will use all features by default
pca = PCA()

# Apply PCA on the standardized data
pca.fit(df1_standardized)

# Retrieve the explained variance ratio for each principal component
explained_variance_ratio = pca.explained_variance_ratio_

# Calculate the cumulative explained variance, which shows how much total variance is explained by each successive principal component
cumulative_variance = explained_variance_ratio.cumsum()

# Plot the relationship between the number of principal components and the cumulative explained variance
plt.figure(figsize=(8, 6))
plt.plot(range(1, len(explained_variance_ratio) + 1), cumulative_variance, marker='o', linestyle='--', color='dodgerblue')
plt.xlabel('Number of Principal Components', fontsize=14, fontweight='bold')  # Label for x-axis with bold font
plt.ylabel('Cumulative Explained Variance', fontsize=14, fontweight='bold')  # Label for y-axis with bold font
plt.title('Cumulative Variance Explained by Principal Components', fontsize=14, fontweight='bold')  # Title of the plot with bold font
plt.grid(True)  # Enable grid for easier reading
plt.show()  # Display the plot

# Output the cumulative explained variance when the number of principal components is 35 (variance threshold)
# Index 34 corresponds to the 35th principal component because indexing starts from 0
threshold_35 = cumulative_variance[34]
print(f'When the number of principal components is 35, the cumulative explained variance is: {threshold_35:.4f}')
```

```python
# Initialize PCA and set the number of components to retain to 38
pca = PCA(n_components=38)

# Apply PCA on the standardized data for dimensionality reduction
df1_pca = pca.fit_transform(df1_standardized)

# Convert the dimensionality-reduced data into a DataFrame and display the first few rows
df1_pca = pd.DataFrame(df1_pca)
df1_pca.head()  # This command will show the first 5 rows of the DataFrame by default
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Using the Elbow Method - Calculate SSE (Sum of Squared Errors) for different numbers of clusters
sse = []  # To store the SSE for each number of clusters
k_range = range(1, 17)  # Test clustering from 1 to 16 clusters

for k in k_range:
    # Initialize KMeans specifying the number of clusters and a fixed random state for reproducibility
    kmeans = KMeans(n_clusters=k, random_state=2024)
    
    # Fit the model on the dimensionality-reduced data (df1_pca)
    kmeans.fit(df1_pca)
    
    # Append the inertia_ attribute (which represents SSE) of the KMeans model to the sse list
    sse.append(kmeans.inertia_)

# Plotting the Elbow Method graph to determine the optimal number of clusters
plt.figure(figsize=(8, 6))  # Set the figure size for better readability
plt.plot(k_range, sse, marker='o', linestyle='--', color='dodgerblue')  # Plot k versus SSE with specified styling
plt.xlabel('Number of Clusters (k)', fontsize=16, fontweight='bold')  # Label for x-axis with bold font
plt.ylabel('SSE (Sum of Squared Errors)', fontsize=16, fontweight='bold')  # Label for y-axis with bold font
plt.title('Elbow Method For Optimal k', fontsize=16, fontweight='bold')  # Title of the plot with bold font
plt.grid(True)  # Enable grid for easier visualization of points
plt.show()  # Display the plot
```

```python
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Using the Silhouette Coefficient method - Calculate silhouette scores for different numbers of clusters
silhouette_scores = []  # To store the silhouette scores for each number of clusters
k_range = range(2, 15)  # Test clustering from 2 to 14 clusters because silhouette score is undefined for a single cluster

for k in k_range:
    # Initialize KMeans specifying the number of clusters and a fixed random state for reproducibility
    kmeans = KMeans(n_clusters=k, random_state=2024)
    
    # Fit the model on the dimensionality-reduced data (df1_pca) and predict the cluster labels
    labels = kmeans.fit_predict(df1_pca)
    
    # Calculate the silhouette score which provides a succinct graphical representation of how well each object lies within its cluster
    silhouette_avg = silhouette_score(df1_pca, labels)
    
    # Append the average silhouette score to the list
    silhouette_scores.append(silhouette_avg)

# Plotting the Silhouette Score graph to determine the optimal number of clusters
plt.figure(figsize=(8, 6))  # Set the figure size for better readability
plt.plot(k_range, silhouette_scores, marker='o', linestyle='--', color='dodgerblue')  # Plot k versus silhouette scores with specified styling
plt.xlabel('Number of Clusters (k)', fontsize=16, fontweight='bold')  # Label for x-axis with bold font
plt.ylabel('Silhouette Score', fontsize=16, fontweight='bold')  # Label for y-axis with bold font
plt.title('Silhouette Score For Optimal k', fontsize=16, fontweight='bold')  # Title of the plot with bold font
plt.grid(True)  # Enable grid for easier visualization of points
plt.show()  # Display the plot
```

```python
# Perform clustering using KMeans, specifying the number of clusters to be 8
kmeans = KMeans(n_clusters=8, random_state=2024)

# Apply KMeans clustering on the dimensionality-reduced data (df1_pca)
kmeans.fit(df1_pca)

# Retrieve the cluster labels for each sample from the fitted KMeans model
labels = kmeans.labels_

# Add a 'Cluster' column to the dimensionality-reduced DataFrame, containing the cluster labels for each sample
df['Cluster'] = labels
df1_pca['Cluster'] = labels  #This data is required for subsequent visualization.
df1_pca['Framework'] = framework_col.values
df
```

## Prediction Results on New Data(Generate structure）

Below, we present the prediction results on newly generated data using the built-in unsupervised learning model. The algorithm is thoroughly described in the main text.

Here, we apply the `kmeans.fit_predict` method from **scikit-learn** to perform the prediction.  
For more details, refer to the [official documentation](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.KMeans.html).


```python
# Import necessary libraries
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

# Read a new CSV file containing the data for analysis
new_file_path = r'C:\Users\lenovo\Desktop\ML-Zeolite-Selective\data\2_unsupervised_learning\2_input\CDVAE_2000_features.csv'
df_new = pd.read_csv(new_file_path, encoding='utf-8')

# Remove the first column (assuming it's non-feature related)
df_new_features = df_new.iloc[:, 1:]

# Standardize the new data to ensure that each feature contributes equally to the distance calculations
scaler_new = StandardScaler()
df_new_standardized = scaler_new.fit_transform(df_new_features)

# Apply dimensionality reduction using the previous PCA model configuration (n_components=38)
pca_new = PCA(n_components=38)
df_new_pca = pca_new.fit_transform(df_new_standardized)

# Use the previously trained KMeans model to predict clusters on the new data
# Ensure consistency by specifying the same parameters used in the original model
kmeans = KMeans(n_clusters=8, random_state=2024)  # Make sure to use the same random state for reproducibility
labels_new = kmeans.fit_predict(df_new_pca)

# Add the cluster labels to the dimensionality-reduced DataFrame
df_new_pca_df = pd.DataFrame(df_new_pca)
df_new_pca_df['Cluster'] = labels_new

# Display the first few rows of the DataFrame with cluster labels
df_new_pca_df.head()
```

```python
# Define the path where the output file will be saved
output_file_path = r'C:\Users\lenovo\Desktop\ML-Zeolite-Selective\data\2_unsupervised_learning\2_output\CDVAE_2000_predict.csv'

# Save the DataFrame containing cluster labels to a CSV file at the specified path
# index=False ensures that the index column is not written into the output file
# encoding='utf-8' specifies the character encoding for the output file, ensuring compatibility across different systems
df_new_pca_df.to_csv(output_file_path, index=False, encoding='utf-8')

# Print a confirmation message indicating the file has been successfully saved
print(f'Cluster prediction results have been successfully saved to {output_file_path}')
```
