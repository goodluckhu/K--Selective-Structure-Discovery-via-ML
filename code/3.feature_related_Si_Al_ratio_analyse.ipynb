```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Define file path
file_path = r'C:\Users\lenovo\Desktop\ML-Zeolite-Selective\data\3_feature_related_Si_Al_ratio_analyse\Train_Descriptor_Si_Al_Ratio.csv'

# Read the CSV file
data = pd.read_csv(file_path)

# Display the data
data
```

```python
from lightgbm import LGBMRegressor
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Extract X and Y
X = data.iloc[:, :-1]  # Select all columns except the last one
Y = data.iloc[:, -1]   # Select the last column

# Initialize LightGBM regressor model
lgbm_model = LGBMRegressor(random_state=42)

# Train the model
lgbm_model.fit(X, Y)

# Get feature importance
feature_importance = lgbm_model.feature_importances_

# Convert feature importance to DataFrame for easier viewing
importance_df = pd.DataFrame({
    'Feature': X.columns,
    'Importance': feature_importance
})

# Sort by importance
importance_df = importance_df.sort_values(by='Importance', ascending=False)

# Output feature importance
print(importance_df)

# Visualize feature importance
plt.figure(figsize=(12, 8), dpi=500)
sns.barplot(x='Importance', y='Feature', data=importance_df, palette='Blues_r')

# Bold all labels
plt.title('Feature Importance of Si/Al Ratios from LightGBM', fontsize=16, fontweight='bold')
plt.xlabel('Importance', fontsize=14, fontweight='bold', labelpad=10)
plt.ylabel('Feature', fontsize=14, fontweight='bold', labelpad=10)

# Set bold font for ticks
plt.xticks(fontsize=12, fontweight='bold')
plt.yticks(fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```
