# Here we demonstrate how to compute component descriptors

## Using the Pymatgen program to read the cif file corresponding to the structure and then compute the component descriptors

```python
from matminer.featurizers.composition.composite import ElementProperty
from pymatgen.io.cif import CifParser
import os
import csv
from tqdm import tqdm
import warnings
```

```python
# Set the CIF file directory and output path
cif_directory = r'C:\Users\lenovo\Desktop\ML-Zeolite-Selective\data\1_descriptor_calculation\1_descriptor_calculation_cif'
output_file = r'C:\Users\lenovo\Desktop\ML-Zeolite-Selective\data\1_descriptor_calculation\1_output\zeolite-composition.csv'
error_output_file = r'C:\Users\lenovo\Desktop\ML-Zeolite-Selective\data\1_descriptor_calculation\1_output\error.txt'
```

```python
warnings.filterwarnings("ignore")
# Create the feature extractor object
ep = ElementProperty(data_source="pymatgen", features=["Z", "atomic_radius_calculated", "van_der_waals_radius", "mendeleev_no", "molar_volume", "thermal_conductivity", "boiling_point", "melting_point", "liquid_range", "X"], stats=["minimum", "maximum", "range", "mean", "std_dev"])

# Initialize a list to store the composition descriptors
all_features = []

# Get the feature labels
feature_labels = ep.feature_labels()

# Create a list to store the composition information
composition_info = []

# Create a list to store the error information
error_info = []

# Iterate over each CIF file in the directory
for filename in tqdm(os.listdir(cif_directory)):
    if filename.endswith('.cif'):
        # Create the CIF file path
        cif_file = os.path.join(cif_directory, filename)

        try:
            # Parse the CIF file and extract the composition data
            parser = CifParser(cif_file)
            structure = parser.parse_structures(primitive=False)[0]
            composition = structure.composition

            # Extract the composition descriptors
            features = ep.featurize(composition)

            # Append the file name and descriptors to the list
            all_features.append([filename] + features)

            # Append the composition information to the list
            composition_info.append("Composition of {}: {}".format(filename, str(composition)))

        except Exception as e:
            error_info.append("Error processing {}: {}".format(filename, str(e)))

# Write the composition descriptors to the output CSV file
with open(output_file, 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)

    # Write the header row
    writer.writerow(['File Name'] + feature_labels)

    # Write the data rows
    writer.writerows(all_features)

# Write the error information to the output text file
with open(error_output_file, 'w') as txtfile:
    txtfile.write('\n'.join(error_info))

print("Composition descriptors have been generated and saved to", output_file)
print("Error information has been saved to", error_output_file)
```
